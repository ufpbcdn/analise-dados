```python
import pandas as pd

# Tabela A
municipio = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'nome': ['São Paulo', 'Rio de Janeiro', 'Belo Horizonte', 'João Pessoa'],
    'uf': ['SP', 'RJ', 'MG', 'PB']})

# Tabela B
populacao = pd.DataFrame({
    'id': [1, 2, 4],
    'populacao': [12300000, 6748000, 817511]})

# Junção de tabelas usando merge
outer = pd.merge(municipio, populacao, on='id', how='outer')
inner = pd.merge(municipio, populacao, on='id', how='inner')
left = pd.merge(municipio, populacao, on='id', how='left')
right = pd.merge(municipio, populacao, on='id', how='right')

print(f"Junção externa (outer join): {len(outer)} registros")
print(f"Junção interna (inner join): {len(inner)} registros")
print(f"Junção à esquerda (left join): {len(left)} registros")
print(f"Junção à direita (right join): {len(right)} registros")
```

```python
file = '~/Downloads/dados/divulgacao_anos_iniciais_municipios_2023/divulgacao_anos_iniciais_municipios_2023.xlsx'
ideb = pd.read_excel(file, skiprows=9, na_values=['-', 'ND', 'N/A', 'N/D', 'N/A', 'ND/NA', 'ND/N/A'])

list(ideb.columns)
selecao_colunas = ['SG_UF', 'CO_MUNICIPIO', 'NO_MUNICIPIO', 'REDE',
                   'VL_OBSERVADO_2005', 'VL_OBSERVADO_2007', 'VL_OBSERVADO_2009','VL_OBSERVADO_2011', 'VL_OBSERVADO_2013', 'VL_OBSERVADO_2015', 'VL_OBSERVADO_2017', 'VL_OBSERVADO_2019', 'VL_OBSERVADO_2021', 'VL_OBSERVADO_2023']

ideb = ideb[selecao_colunas]

print("Antes da transformação:")
display(ideb.head())

ideb.info()

# Transformando a tabela de formato largo para formato longo
ideb = (
    ideb.melt(id_vars=['SG_UF', 'CO_MUNICIPIO', 'NO_MUNICIPIO', 'REDE'],
          value_vars=['VL_OBSERVADO_2005', 'VL_OBSERVADO_2007', 'VL_OBSERVADO_2009','VL_OBSERVADO_2011', 'VL_OBSERVADO_2013', 'VL_OBSERVADO_2015', 'VL_OBSERVADO_2017', 'VL_OBSERVADO_2019', 'VL_OBSERVADO_2021', 'VL_OBSERVADO_2023'],
          var_name='ano', value_name='ideb'
          )
          .assign(ano=lambda x: x['ano'].str.replace('VL_OBSERVADO_', '').astype(int))
          .rename(columns=lambda x: x.lower())
)

print("Depois da transformação:")
display(ideb.head())
```

## Formato de Dados

- WIDE: os dados se expandem em colunas
- LONG: os dados se expandem em linhas (Do ponto de vista estatístico, o LONG seria a melhor escolha)

```python
from unidecode import unidecode
# DTB
file = '~/Downloads/dados/DTB_2025/RELATORIO_DTB_BRASIL_2025_MUNICIPIOS.xls'
dtb = pd.read_excel(file, skiprows=6)

# Normalização dos nomes das colunas
dtb.rename(columns=lambda x: unidecode(x.strip().lower()
                                       .replace(' ', '_')
                                       .replace('nome_', 'no_')
                                       .replace('codigo_', 'co_')), inplace=True)

dtb = dtb.rename(columns={'codigo_municipio_completo': 'co_municipio'})
dtb.columns
```

```python
# Sociais
variaveis = ['ANO', 'CODMUN7', 'PPOB', 'T_ANALF18M', 'GINI',
             'RDPC', 'IDHM', 'POP']
file = '~/Downloads/dados/censo_total_1991_2010.xlsx'
atlas = (pd.read_excel(file, sheet_name=1)
         .rename(columns=lambda x: x.upper())
         [variaveis])

```

```python
# Padronizar o nome das colunas
atlas.rename(columns={'CODMUN7': 'co_municipio'}, inplace=True)
atlas.rename(columns=lambda x: x.lower(), inplace=True)
atlas.ano.value_counts()
```

```python
print("Ideb", ideb.shape)
print("DTB", dtb.shape)

ideb_r = ideb.drop(columns=['sg_uf', 'no_municipio'])
atlas_r = atlas.query('ano == 2010').drop(columns=['ano'])

df = (
    dtb
    .assign(co_regiao = lambda x: x.co_municipio.astype(str).str[:1].astype(int),
            no_regiao = lambda x: x.co_regiao.map({1: 'Norte', 2: 'Nordeste', 3: 'Sudeste', 4: 'Sul', 5: 'Centro-Oeste'}))
    .merge(ideb_r, on='co_municipio', how='inner', validate='1:m')
    .merge(atlas_r, on='co_municipio', how='left')
)
print("DF", df.shape)
df
```
