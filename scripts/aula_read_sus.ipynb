```python
file = '/Users/alessiotony/Downloads/licitacoes-2025.csv'

import pandas as pd

df = pd.read_csv(file, sep=';', 
                 decimal=',', 
                 thousands='.', 
                 dtype={'cpf_cnpj_proponente': str},
                 parse_dates=['data_homologacao'], 
                 date_format='%d/%m/%Y')

df.info()
df['pessoa_fisica'] = df['cpf_cnpj_proponente'].str.len() == 11

# Remover espaços extras em variáveis textuais
cols_texto = df.select_dtypes(include="object").columns

for col in cols_texto:
    df[col] = df[col].astype(str).str.strip()

# ------------------------------------------------------------
# 2. Visão geral da base
# ------------------------------------------------------------

print("=" * 80)
print("1. VISÃO GERAL DA BASE")
print("=" * 80)

print(f"Arquivo: {file}")
print(f"Número de linhas: {df.shape[0]:,}".replace(",", "."))
print(f"Número de colunas: {df.shape[1]}")

display(df.head())

print("\nInformações gerais:")
df.info()
```

```python
# ------------------------------------------------------------
# 3. Dicionário inicial das variáveis
# ------------------------------------------------------------

diagnostico_colunas = pd.DataFrame({
    "coluna": df.columns,
    "tipo": df.dtypes.astype(str).values,
    "n_observacoes": df.notna().sum().values,
    "n_ausentes": df.isna().sum().values,
    "perc_ausentes": (df.isna().mean() * 100).round(2).values,
    "n_valores_unicos": df.nunique(dropna=True).values
})

diagnostico_colunas = diagnostico_colunas.sort_values(
    by=["perc_ausentes", "n_valores_unicos"],
    ascending=[False, True]
)

display(diagnostico_colunas)
```

```python
# ------------------------------------------------------------
# 4. Diagnóstico de duplicidades
# ------------------------------------------------------------

print("=" * 80)
print("2. DUPLICIDADES")
print("=" * 80)

n_duplicadas_exatas = df.duplicated().sum()

print(f"Linhas exatamente duplicadas: {n_duplicadas_exatas:,}".replace(",", "."))

# Possível chave de identificação da proposta
chave_proposta = [
    "codigo_unidade_gestora",
    "numero_licitacao",
    "numero_protocolo_tce",
    "ano_licitacao",
    "cpf_cnpj_proponente",
    "valor_ofertado"
]

colunas_chave_existentes = [col for col in chave_proposta if col in df.columns]

duplicadas_chave = df.duplicated(subset=colunas_chave_existentes).sum()

print(
    f"Duplicidades pela chave analítica sugerida: "
    f"{duplicadas_chave:,}".replace(",", ".")
)

if duplicadas_chave > 0:
    display(
        df[df.duplicated(subset=colunas_chave_existentes, keep=False)]
        .sort_values(colunas_chave_existentes)
        .head(20)
    )
```

```python
# ------------------------------------------------------------
# 5. Diagnóstico de datas
# ------------------------------------------------------------

print("=" * 80)
print("3. DIAGNÓSTICO DE DATAS")
print("=" * 80)

print(f"Datas inválidas ou não convertidas: {df['data_homologacao'].isna().sum():,}".replace(",", "."))

print("\nPeríodo observado:")
print(f"Data mínima: {df['data_homologacao'].min()}")
print(f"Data máxima: {df['data_homologacao'].max()}")

# Extração de componentes temporais
df["ano_homologacao"] = df["data_homologacao"].dt.year
df["mes_homologacao"] = df["data_homologacao"].dt.month
df["ano_mes_homologacao"] = df["data_homologacao"].dt.to_period("M").astype(str)

# Inconsistência entre ano_licitacao e ano da homologação
df["ano_licitacao_diferente_homologacao"] = (
    df["ano_licitacao"] != df["ano_homologacao"]
)

print(
    "\nRegistros em que o ano da licitação difere do ano da homologação:",
    f"{df['ano_licitacao_diferente_homologacao'].sum():,}".replace(",", ".")
)

display(
    df.loc[
        df["ano_licitacao_diferente_homologacao"],
        [
            "nome_municipio",
            "codigo_unidade_gestora",
            "numero_licitacao",
            "ano_licitacao",
            "data_homologacao",
            "ano_homologacao",
            "modalidade",
            "valor_ofertado"
        ]
    ].head(20)
)
```

### Conclusão sobre a qualidade dos dados e potenciais limitações

A base de licitações de 2025 apresenta, em termos gerais, boa completude e consistência estrutural inicial. Todas as variáveis analisadas possuem 49.909 observações preenchidas, sem registros ausentes identificados. As variáveis centrais para identificação, localização, modalidade, data, proponente, valor ofertado e situação da proposta foram importadas com tipos compatíveis com a análise, incluindo a conversão adequada da variável `data_homologacao` para formato de data. Também não foram identificadas inconsistências entre o ano da licitação e o ano da homologação, uma vez que todos os registros pertencem ao exercício de 2025.

Apesar dessa boa qualidade aparente, foram identificadas três linhas exatamente duplicadas, também coincidentes com a chave analítica sugerida. Como o volume é muito pequeno em relação ao total da base, esse problema não compromete a análise exploratória geral, mas recomenda-se remover ou investigar esses registros antes de análises agregadas, especialmente quando o objetivo for estimar valores totais, número de propostas ou concentração por fornecedor.

A principal limitação observada está relacionada à variável `valor_ofertado`. Embora não existam valores ausentes, negativos ou inconsistentes em termos formais, a distribuição apresenta forte assimetria e valores extremamente elevados. A média do valor ofertado é muito superior à mediana, indicando que poucos registros de grande magnitude influenciam fortemente os indicadores agregados. O maior valor observado ultrapassa R$ 25 bilhões, valor incompatível com a escala usual de licitações municipais, especialmente considerando o objeto associado a aquisição de material de construção. Outros valores também aparecem na casa de centenas de milhões ou bilhões, o que sugere possível problema de registro, erro de separador decimal, erro de digitação, unidade monetária inadequada ou lançamento indevido do valor global/estimado.

Essa característica impõe cautela na interpretação de totais, médias e rankings por município, unidade gestora, modalidade ou fornecedor. Indicadores baseados na soma ou na média de `valor_ofertado` podem ser fortemente distorcidos por poucos registros extremos. Para análises descritivas mais robustas, recomenda-se utilizar também mediana, percentis, análise com e sem outliers e validação manual dos maiores valores. Em especial, os maiores registros devem ser confrontados com os documentos originais ou com a fonte oficial de consulta do TCE antes de qualquer conclusão substantiva.

Outra limitação relevante é que a variável `situacao_proposta` possui apenas um valor observado, “Vencedora”. Isso indica que a base provavelmente contém apenas propostas vencedoras ou registros já filtrados, não permitindo avaliar competição entre participantes, número de propostas por certame, diferença entre valores ofertados e vencedores, descontos obtidos ou intensidade concorrencial. Portanto, a unidade de análise mais adequada parece ser a proposta vencedora vinculada a uma licitação/homologação, e não o processo competitivo completo.

Também se observa que `ano_licitacao` possui apenas o valor 2025, o que é esperado para uma base anual, mas limita análises temporais comparativas entre exercícios. A dimensão temporal disponível permite análise mensal dentro de 2025, mas não permite inferir tendências históricas, sazonalidade interanual ou mudanças de padrão em relação a anos anteriores.

Por fim, variáveis textuais como `objeto_licitacao`, `nome_proponente` e `descricao_unidade_gestora` apresentam elevada cardinalidade, o que é esperado nesse tipo de base, mas pode conter variações de grafia, abreviações, inconsistências de padronização e descrições semelhantes escritas de formas diferentes. Para análises mais avançadas, recomenda-se aplicar procedimentos de normalização textual, categorização de objetos e padronização de fornecedores, especialmente quando o objetivo for identificar padrões de concentração, recorrência de fornecedores ou tipos de contratação.

Em síntese, a base é adequada para uma análise exploratória inicial sobre distribuição territorial, modalidades, unidades gestoras, fornecedores e valores homologados em 2025. No entanto, antes de produzir conclusões gerenciais ou indicadores oficiais, é necessário tratar duplicidades, investigar valores extremos, confirmar a interpretação da variável `valor_ofertado` e reconhecer que a base parece representar apenas propostas vencedoras, não o conjunto completo do processo licitatório.
